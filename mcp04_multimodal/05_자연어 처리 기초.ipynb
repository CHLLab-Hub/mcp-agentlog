# **1. 자연어 처리(NLP) 기초**

자연어 처리(Natural Language Processing, NLP)는 **컴퓨터가 사람이 사용하는 언어을 처리하고, 이해하고, 생성할 수 있도록 만드는 인공지능 분야**이다. 과거에는 단어의 빈도와 통계적인 특징을 이용하는 방법이 많이 사용됐지만, 현재는 transformer와 대규모 언어 모델(LLM)을 중심으로 자연어 처리가 발전하고 있다.
```
Text > Tokenization > Token > Token ID(중복 Token은 같은 ID 부여) > Padding / Attention Mask() > Embedding > Neural Network(train) > Transformer > LLM
```

````python
import torch
````

````python
text = "나는 자연어를 공부하고 있어요."
vocab = {"<PAD>": 0, "<UNK>": 1, "나는": 2, "자연어를":3, "공부하고":4, "있어요.":5}

tokens = text.split()
token_ids = [vocab.get(token, vocab["<UNK>"]) for token in tokens]

input_ids = torch.tensor(token_ids, dtype=torch.long)

# 여기서 2, 3, 4, 5는 의미의 크기가 아니라 단순한 token 식별 번호
# 이후에 embedding이 각 번호를 실수 vector로 바꾸고 학습시킴
print("원문: ", text)
print("Token: ", tokens)
print("Token ID: ", token_ids)
print("PyTorch Tensor: ", input_ids)
print("dtype: ", input_ids.dtype)
````

### **1. 자연어 처리란?**
컴퓨터가 자연어를 처리하는 기술을 자연어 처리라고 한다. 대표적인 자연어 처리 작업에는 아래와 같은 것들이 있다.
- 텍스트 분류
- 감정 분석
- 문서 요약
- 기계 번역
- 질의 응답
- 정보 검색
- 챗봇
- 문장 생성
- RAG
- LLM 기반 Agent
> 최근에는 자연어 처리 기술이 이미지, 음성, 영상 등의 데이터와 결합되면서 multimodal AI로 확장

### **2. 컴퓨터는 문장을 그대로 이해하지 못함**
컴퓨터의 신경망 모델은 문자열 자체를 그대로 계산하지 않는다. 자연어를 모델에 입력하기 위해서는 숫자로 변환하는 과정이 필요하다.
```
"오늘 날씨가 정말 좋다" > Tokenization > Token > Token ID > Embedding > Vector > Neural Network
```
> 자연어 처리에서 가장 중요한 시작점은 결국 "문장을 어떻게 나누고, 어떻게 숫자로 표현할 것인가?"이다.

### **3. Corpus와 Document**
1. Corpus
    
    Corpus(말뭉치)는 자연어 처리에서 분석하거나 학습하기 위해 수집한 전체 텍스트 데이터의 집합이다.
2. Document
    
    Corpus를 구성하는 각각의 텍스트 데이터이다.
    - Document라고 해서 반드시 긴 문서를 의미하는 것은 아니다.
    - 예시) 뉴스, 기사 하나, 리뷰 하나, 댓글하나 , 문장 하나

### **4. Tokenization**
텍스트를 모델이 처리할 수 있는 작은 단위인 token으로 나누는 과정이다. token은 반드시 단어 하나를 의미하지 않는다. 
- 예시) 문장, 단어, 형태소, 문자, subword
> 과거에는 영어에서 공백을 기준으로 단어를 나누거나 한국어에서는 형태소 분석기를 이용하는 방법을 많이 사용했으나, 현대의 transformer 기반 NLP에서는 subword tokenization이 매우 중요하다.

# **2. Hugging Face**
자연어 처리뿐 아니라 이미지, 음성, 영상, multimodal AI 모델을 공유하고 사용하는데 필요한 여러 도구와 서비스를 제공하는 AI 생태계이자 회사명 

````python
from transformers import AutoTokenizer # Hugging Face에서 제공
````

````python
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base") # from_pretrained(): Hugging Face hub의 모델에서 tokenizer 설정과 vocabulary를 가져옴

text = "나는 자연어를 공부하고 있어요."
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Token:", tokens)
print("Token ID:", token_ids)
````

### **1. Vocabulary**
Vocabulary는 tokenizer가 사용하는 tokens의 집합이다. 각 token은 고유한 정수 ID가 할당된다.
- Token ID 자체는 언어적 의미가 없다.

````python
# [PAD], [UNK], [CLS], [SEP]와 같은 특수한 목적을 가진 token을 special token이라고 함
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base")

print("Vocabulary 크기:", tokenizer.vocab_size)
print("[PAD] token ID:", tokenizer.pad_token_id)
print("[UNK] token ID:", tokenizer.unk_token_id)
print("[CLS] token ID:", tokenizer.cls_token_id) # classification token: 문장 분류
print("[SEP] token ID:", tokenizer.sep_token_id) # separation token: 문장 구분
````

### **2. OOV**
OOV(Out-Of-Vocabulary)는 vocabulary에 존재하지 않는 token이 입력되는 문제를 일컫는다. 과거의 단어 단위 vacabulary에서는 처음 보는 단어를 <UNK>와 같은 unknown token으로 처리하는 경우가 많았다. 현대 NLP에서는 subword tokenization을 이용하여 이와같은 문제를 크게 줄였다.
> 단어 단위 vocabulary의 OOV 문제는 subword tokenizer로 처음 보는 단어를 더 작은 조각으로 나눠서 OOV를 크게 줄일 수 있다.

````python
vocab = {"<PAD>": 0, "<UNK>": 1, "나는":2, "파이토치를":3, "공부한다.":4}
tokens = ['나는', '트랜스포머를', '공부한다.']
token_ids = [vocab.get(token, vocab["<UNK>"]) for token in tokens]

print(token_ids)
````

### **3. Token ID**
Tokenization이 끝나면 각각의 token을 vocabulary에 정의된 정수로 ID를 반환한다.
```
"오늘 날씨가 좋다." > Tokenization > [Token1, Token2, Token3, Token4, ...] > Token ID > [101, 5662, 4943, 4944, ...]
```

````python
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base")
# return_tensors: 결과를 PyTorch tensor로 반환
# padding: Batch 내에서 가장 긴 문장의 길이에 맞추기 위해 padding token을 추가
# truncation: 문장 길이가 모델의 max_length보다 길면 자르기
encoded = tokenizer("나는 자연어를 공부하고 있어요.", return_tensors="pt", padding=True, truncation=True)

# Token과 special token을 정수 ID로 바꾼 tensor
print(encoded["input_ids"])
# 실제 token은 1, padding 위치는 0으로 표시된 tensor >> 0은 padding으로 채워진 의미 없는 것
print(encoded["attention_mask"])
print(type(encoded["input_ids"]))
print(encoded["input_ids"].dtype)
# token_type_ids: 일부 BERT 모델에서 문장 A와 문장 B를 구분하기 위해 사용되는 tensor, 모든 모델이 반환하는 것은 아님

# encoded: 단순한 tensor 하나가 아니라 모델 입력에 필요한 여러값을 key로 담은 BatchEncoding 객체
````

### **4. Padding**
문장은 각각 길이가 다르기 때문에 tensor의 크기를 맞추기 위해 짧은 squence에 특별한 padding token을 추가할 수 있다.
```
문장 1
[15, 23, 51]

문장 2
[73, 18, 94, 21, 55]

Padding 후
[15, 23, 51, 0, 0]
[73, 18, 94, 21, 55]
```
> 0을 \<PAD\>의 ID로 사용했다.

````python
from torch.nn.utils.rnn import pad_sequence
````

````python
sentence_A = torch.tensor([15, 23, 51], dtype=torch.long)
sentence_B = torch.tensor([73, 18, 94, 21, 55], dtype=torch.long)

padded = pad_sequence([sentence_A, sentence_B], padding_value=0, batch_first=True) # batch_first=True: 실제 문장 형태로 묶이게 됨

print(padded)
print('shape:', padded.shape)
````

### **5. Padding과 Attention mask**
Padding을 추가하면 문제가 생길 수 있다. 모델 입장에서 Padding도 하나의 입력 숫자이므로 실제 문장이 아님에도 학습에 이용한다. 따라서, Transformer 모델에서는 Attention mask라는 것을 사용한다. Attention mask는 실제 문장 위치는 1, padding 위치는 0으로 표시된 tensor이다. 모델은 Attention mask를 이용해 padding 위치를 무시하고 실제 문장 위치만 학습에 이용한다.에서는 attention mask를 함게 사용한다.
```
Input IDs
[15, 23, 51, 0, 0]

Attention mask
[1, 1, 1, 0, 0]

1: 실제 token, 0: padding
```

````python
input_ids = torch.tensor([
    [15, 23, 51, 0, 0],
    [73, 18, 94, 21, 55],
], dtype=torch.long)

attention_mask = (input_ids != 0).long()
print(attention_mask)
````

# **3. 벡터화**

벡터화는 텍스트 데이터를 숫자 형태로 변환하여 머신러닝 또는 딥러닝 모델에서 처리할 수 있도록 만드는 과정이다. 이를 위해 단어의 빈도를 기반으로 한 Bag-of-Words(BOW), TF-IDF와 같은 방법부터, 단어 간의 의미적 관계를 학습하는 word embedding(단어 임베딩), 문장의 문맥과 구조를 반영하는 transformer 기반 임베딩(BERT, GPT) 등 다양한 기법이 사용된다. 벡터화는 단어, 문장, 문서와 같은 텍스트 데이터의 특징을 수치적으로 표현하며, 단순히 단어의 빈도를 반영하거나 의미적 유사성을 학습하여 벡터 공간에서 단어 간 관계를 나타낸다. 데이터의 특성과 분석 목적에 따라 적절한 벡터화 기법을 선택하는 것이 중요하다.

### **1. 원-핫 인코딩**
원-핫 인코딩(One-Hot Encoding)은 텍스트 데이터를 숫자 벡터로 변환하는 가장 단순한 방법 중 하나로, 단어 집합(vocabulary)에 있는 단어들을 고유한 숫자 벡터로 표현한다. 이 방법은 각 단어를 벡터의 한 차원으로 지정하고, 해당 단어에만 1을 넣고 나머지는 0으로 설정한다. 이를 통해 텍스트 데이터를 컴퓨터가 이해할 수 있는 숫자로 변환한다. 예를 들어, 단어 집합이 ["I", "love", "Python"]이라면, "love"는 [0, 1, 0]으로 표현된다. 이 방법은 단어를 간단하고 직관적으로 표현할 수 있지만, 단어 간의 의미적 관계를 반영하지 못하며, 단어 집합 크기만큼 고차원 희소 벡터를 생성하므로 메모리 낭비가 크다.

![원-핫 인코딩](./images/원-핫%20인코딩.png)

````python
import torch.nn.functional as F
````

````python
token_id = torch.tensor(2)
vocab_size = 5

one_hot = F.one_hot(token_id, num_classes=vocab_size)

print(one_hot)
print(one_hot.shape)
````

### ※ 희소 벡터
희소 벡터(Sparse Vector)는 대부분의 값이 0인 벡터를 의미하며, 자연어 처리에서 원-핫 인코딩(One-Hot Encoding)과 같은 방법으로 단어를 표현할 때 자주 나타난다. 예를 들어, 단어 집합의 크기가 10,000이라면 각 단어를 표현하는 벡터는 10,000차원이 되고, 이 중 단 하나의 위치만 1이고 나머지는 모두 0으로 채워진다. 이러한 벡터는 메모리 낭비와 계산 복잡도를 초래하며, 특히 단어 집합이 클수록 문제가 심각해진다. 이를 해결하기 위해 단어를 저차원의 조밀한 벡터(Dense Vector)로 변환하는 워드 임베딩(Word Embedding) 기법이 자주 사용된다.

### **2. Bag of Words**
Bag of Words(BOW)는 텍스트 데이터를 벡터화하는 간단한 방법으로, 문장을 구성하는 단어들의 등장 빈도를 기반으로 표현한다. 각 문장은 단어 집합(vocabulary)의 크기만큼 차원을 가지며, 단어가 해당 문장에 몇 번 나타나는지 그 빈도를 벡터의 값으로 나타낸다. 예를 들어, 단어 집합이 ["I", "love", "Python"]이고 문장이 "I love Python Python"이라면, 벡터는 [1, 1, 2]로 표현된다. 단어의 순서나 문맥은 고려되지 않고 단순히 단어의 빈도만 반영되기 때문에, 문장의 구조나 의미를 표현하는 데는 한계가 있다. 하지만 구현이 간단하고 텍스트 데이터를 빠르게 벡터화할 수 있어 자연어 처리의 기초적인 방법으로 널리 사용된다. TF-IDF와 같은 확장 기법을 통해 BOW의 단점을 보완하기도 한다.

### **3. TF-IDF**
TF-IDF(Term Frequency-Inverse Document Frequency)는 텍스트 데이터에서 단어의 중요도를 측정하는 방법으로, 단어의 빈도(TF)와 역문서 빈도(IDF)를 조합하여 계산된다. 단어 빈도(TF)는 특정 문서에서 해당 단어가 얼마나 자주 등장하는지를 나타내고, 역문서 빈도(IDF)는 해당 단어가 다른 문서들에 얼마나 흔하게 등장하는지를 반영하여 중요도를 조정한다. 이를 통해, 특정 문서에서 자주 등장하면서도 다른 문서에서는 잘 나타나지 않는 단어의 가중치가 높아진다. 예를 들어, 일반적으로 모든 문서에서 자주 등장하는 단어(예: "the", "is")는 낮은 가중치를 받고, 특정 문서에서만 중요한 단어는 높은 가중치를 받는다. TF-IDF는 문서의 내용을 잘 나타내는 단어를 식별하고, 텍스트 데이터의 벡터화를 통해 검색 엔진, 텍스트 분류 등의 작업에서 널리 사용된다.

![TF_IDF 계산 방법](./images/TF-IDF%20계산%20방법.png)

> 현재 transformer 중심의 NLP를 공부한다면 TF-IDF 공식을 직접 구현하거나 수학적으로 깊게 공부할 필요는 없다.
> 다만, 전톡적인 텍스트 검색과 머신러닝 기반 텍스트 분류에서 사용하는 방밥이라는 개념은 알아두면 좋다.

### **4. Word Embedding**
워드 임베딩(Word Embedding)은 단어를 고차원의 희소 벡터로 표현하는 기존 방식(원-핫 인코딩) 대신, 단어의 의미를 저차원의 밀집 벡터(dense vector)로 변환하는 자연어 처리 기법이다. 이를 통해 단어 간 유사성과 관계를 벡터 공간에 효율적으로 나타낼 수 있으며, 벡터 간의 거리 또는 방향을 통해 단어의 문맥적 의미를 학습한다. 대표적인 워드 임베딩 알고리즘으로는 Embedding Layer, Word2Vec, GloVe, FastText 등이 있으며, 이를 사용하면 언어 모델이 문맥을 이해하거나 추론하는 데 필요한 기초적인 언어적 의미를 학습할 수 있다.
```
Token ID > Embedding > [0.21, -0.72, 0.13, 0.84, ...]
```
> Embedding의 중요한 복적은 token을 신경망이 학습하기 좋은 연속적으로 벡터공간으로 표현하는 것이다.

````python
import torch.nn as nn
````

````python
# num_embeddings: vocabulary의 크기
# embedding_dim: token 하나를 표현할 vector 차원
# padding_idx: padding token의 ID
embedding = nn.Embedding(num_embeddings=1000, embedding_dim=128, padding_idx=0)
print(embedding)

# Embedding layer 내부에는 voca_size * embedding_dim(10000 * 128), 1만 개의 token 각각에는 128차원의 vector가 존재한다.
print("Embedding weight shape:", embedding.weight.shape)
````

### nn.Embedding
Token ID에 해당하는 vector를 embedding table에서 가져오는 lookup table이다.
```
1. Token ID
[15, 27, 103]

2. Embedding Table
15번 vector
27번 vector
103번 vector

3. Embedding vector
```
> 예를 들어, 입력 tensor의 크기가 (batch_size, squence_len)이고 embedding_dim=128이라면 embedding을 통과한 후에는 (batch_size, sequence_len, 128)의 크기가 된다. 
> 이 Shape 변화는 Transformer를 공부할 때 매우 중요하다.

````python
input_ids = torch.tensor([
    [1, 2, 3],
    [4, 2, 5]
], dtype=torch.long)

embedding = nn.Embedding(num_embeddings=10, embedding_dim=4, padding_idx=0)
output = embedding(input_ids)

print('입력 shape:', input_ids.shape)
print('출력 shape:', output.shape)
print(output)
````

- Embedding vector는 사람이 직접 의미를 입력하는 것이 아니다.
- 처음에는 초기값으로 시작하고 모델 학습 과정에서 다른 weight들과 함꼐 업데이트 된다.
```
Embedding > Model > Prediction > Loss > Backpropagation > Embedding weight update
```
> 따라서, nn.Embedding의 weight도 학습 가능한 parameter이다.

````python
embedding = nn.Embedding(num_embeddings=100, embedding_dim=16, padding_idx=0)

print('Weight shape:', embedding.weight.shape)
print('학습 여부:', embedding.weight.requires_grad)
print('Parameter 수:', embedding.weight.numel())
````

# **4. Word2Vec**
Word2Vec은 단어를 벡터로 표현하기 위해 개발된 자연어 처리 기법으로, 단어 간의 의미적 유사성을 수치적으로 나타낼 수 있는 임베딩 벡터를 생성한다. 주로 CBOW(Continuous Bag of Words)와 Skip-gram이라는 두 가지 모델 구조를 사용하며, CBOW는 주변 단어들을 기반으로 중심 단어를 예측하고, Skip-gram은 중심 단어를 기반으로 주변 단어들을 예측하는 방식이다. 이 과정을 통해 학습된 임베딩 벡터는 단어 간의 문맥적 관계를 반영하며, 벡터 공간에서 유사한 의미를 가진 단어들이 서로 가깝게 위치한다. Word2Vec은 대규모 코퍼스를 활용해 고성능 임베딩을 생성할 수 있으며, 자연어 처리 작업에서 사전 훈련된 임베딩으로 널리 사용된다. 벡터가 된 단어들은 연산이 가능하다.
```
한국에서 서울이라는 관계를 빼고, 그 자리에 도쿄라는 관계를 넣으면 어떤 단어가 가까울까?
(한국의 벡터 - 서울의 벡터 + 도쿄의 벡터)
한국-서울+도쿄 > 일본/Noun
```
> Word2Vec는 단어를 그냥 번호로 바꾸는 게 아니라, 단어의 의미와 관계가 어느정도 반영된 숫자의 벡터로 바꾼다.
> 그래서 단어끼리 덧셈, 뺄셈 연산이 가능하다.

### **1. CBOW**
CBOW(Continuous Bag of Words)는 자연어 처리에서 단어 임베딩을 학습하기 위해 사용되는 Word2Vec 알고리즘의 한 방법이다. 이 모델은 주어진 문맥(즉, 중심 단어 주변의 단어들)을 기반으로 중심 단어를 예측하는 방식으로 작동한다. 예를 들어, 문장에서 특정 단어의 좌우 몇 개 단어를 입력으로 받아 해당 중심 단어를 출력으로 예측하는 구조이다. 이를 통해 단어 간의 문맥적 관계를 효과적으로 학습할 수 있으며, 단어를 고차원 공간의 벡터로 표현하여 의미적으로 유사한 단어들이 가까운 위치에 있도록 학습된다. CBOW는 연산적으로 효율적이며, 대규모 데이터에서 빠르고 안정적으로 임베딩을 생성할 수 있다는 장점이 있다.
> 모델은 문맥 단어를 보고 중심 단어를 정확히 예측할 수 있도록 가중치를 학습한다.
> 또한, 단어를 임베딩 벡터로 변환해 의미적으로 유사한 단어들이 가까운 위치로 배치되도록 한다.

### **2. Skip-gram**
Skip-gram은 Word2Vec 알고리즘의 한 방법으로, 주어진 중심 단어로부터 주변 문맥 단어들을 예측하는 방식으로 작동한다. 즉, 중심 단어를 입력으로 사용하고, 그 단어를 기준으로 설정된 윈도우 크기 내에 있는 주변 단어들을 출력으로 예측한다. 예를 들어, 문장에서 "고양이"라는 단어가 중심 단어라면, 그 주변 단어들(예: "귀여운", "자는")을 예측하는 식이다. Skip-gram은 희소한 데이터에서도 성능이 우수하며, 특히 드문 단어의 문맥적 의미를 학습하는 데 효과적이다. 이 모델은 단어의 의미적 관계를 더 잘 반영한 고품질의 단어 벡터를 생성할 수 있지만, CBOW에 비해 계산량이 더 많다는 특징이 있다.
> 해당 모델은 중심 단에서 문맥 단어를 잘 예측하도록 학습되며, 단어 간의 관계를 반영한 임베딩 벡터를 생성한다.
> 학습이 끝나면 의미가 비슷한 단어끼리 가까운 벡터 공간에 베치된다.

### **3. SGNS**
SGNS(Skip-Gram with Negative Sampling)는 Skip-gram 모델의 변형으로, 단어 임베딩을 효율적으로 학습하기 위해 네거티브 샘플링(Negative Sampling) 기법을 사용한다. 기본 Skip-gram은 중심 단어로 주변의 모든 문맥 단어를 예측해야 하지만, SGNS는 주변 단어와의 관계를 일부 샘플로만 학습하여 계산 비용을 줄인다. 즉, 중심 단어와 실제 문맥 단어 쌍(정답)을 학습하면서, 무작위로 선택된 단어(네거티브 샘플)들과의 관계를 "연관성이 없다"고 학습시킨다. 이를 통해 모델은 중심 단어와 문맥 단어 간의 의미적 관계를 효과적으로 학습하면서도 연산량을 크게 줄일 수 있다. SGNS는 Word2Vec 알고리즘에서 널리 사용되며, 실제로 가장 보편적인 단어 임베딩 학습 방법으로 자리 잡았다.
> SGNS는 중심 단어와 문맥 단어의 의미적 관계를 학습하며, positive pair는 중심 단어와 문맥 단어의 임베딩 벡터를 가까운 위치에 배치하고 negative pair는 중심 단어와 연관없는 단어의 벡터를 멀리 배치한다.

### **4. FastText**
FastText는 Facebook AI Research(FAIR)에서 개발한 단어 임베딩 및 텍스트 분류 모델로, Word2Vec과 유사하지만 서브워드(subword) 정보를 활용하여 더 강력한 성능을 제공한다. 기존의 단어 임베딩 기법들은 단어 단위로 벡터를 학습하지만, FastText는 단어를 여러 개의 n-그램 문자 조각(character n-grams)으로 분해하여 학습하기 때문에 희귀 단어(OOV, Out-of-Vocabulary) 처리와 형태소 기반 언어(예: 한국어)에서 뛰어난 성능을 보인다. 또한, FastText는 텍스트 분류에도 활용 가능하며, 빠른 속도와 높은 성능 덕분에 감성 분석, 문서 분류, 검색 시스템 등 다양한 자연어 처리(NLP) 작업에서 널리 사용된다.

##### 영어에서의 처리(예시 중심)
보통 공백 토큰화 + 소문자화를 사용한다.
- n‑그램 길이: 보편적으로 3~6(minn=3, maxn=6)
- 예시: 단어 playing → 문자 3‑그램 일부: <pl, pla, lay, ayi, yin, ing, ng>
    - 접두/접미 경계 표시는 fastText 구현에서 자동으로 붙여준다(예시: <와 >).
- 효과: play, played, playing, playful, playfulness는 많은 서브워드를 공유하므로 의미가 가까운 벡터가 됨
- OOV 예시: 학습에 없던 playfulness도 play, lay, ful, ness 같은 서브워드로 벡터를 구성하므로 유사 단어 검색이 가능함

##### 한국어에서의 처리(포인트와 예시)
한국어는 교착어라 어미·조사가 붙어 형태가 다양하다(예시: 행복, 행복하다, 행복했다, 행복하게).
- fastText는 문자 단위 n‑그램을 쓰므로, 형태가 달라도 공통 부분(“행복”, “강아지”)을 공유하면 유사한 벡터로 묶인다.
- 토큰화 선택지
    - 간단: 띄어쓰기 기준 토큰화만 사용해도 subword 덕분에 어느 정도 해결됨
    - 더 정확: 형태소 분석기(예: MeCab, Okt)로 ‘강아지/들+을’, ‘행복/하+다’처럼 쪼개면 품사·어미를 더 잘 반영함
    - n‑그램 길이 추천: 한글 음절 단위 특성상 min_n=2~3, max_n=4~6 범위를 시도(코퍼스와 길이에 따라 튜닝)

##### 한국어에서의 자모 단위 n-그램
한국어 문자는 사실 초성(ㄱ/ㄴ/ㄷ…), 중성(ㅏ/ㅑ…), 종성(ㄱ/ㄴ/…)이 합쳐진 한글 음절 블록이다.
- fastText는 이를 특별히 분해하지 않고 유니코드 음절 단위(가~힣) 그대로 n-그램을 만든다.
- 다만, 어떤 토크나이저/전처리기를 쓰느냐에 따라 자모 분해(normalize NFD)를 먼저 적용하면, 그 상태에서 fastText가 자모 단위 n-그램을 만드는 효과가 난다.
- 유니코드 NFC(일반 표기): U+AC15 = "강" (1개 문자)
- NFD 정규화 후: ᄀ + ᅡ + ᆼ (자모 3개) → 이 경우 fastText는 "ᄀ", "ᅡ", "ᆼ"을 각각 문자로 보고 n-그램을 만들게 됨
