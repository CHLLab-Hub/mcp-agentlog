# **1. ViT(Vision Transformer)**

ViT(Vision Transformer)는 이미지를 일정 크기의 패치(예: 16×16)로 나눈 뒤 각 패치를 임베딩 벡터로 투영해 토큰 시퀀스로 만들고, 여기에 위치 임베딩을 더해 트랜스포머 인코더(멀티헤드 자기어텐션+FFN)로 처리하여 분류 등의 다운스트림 작업을 수행하는 모델이다. 분류의 경우 BERT처럼 맨 앞에 [CLS] 토큰을 두고 그 출력으로 최종 예측을 한다. CNN이 지역적 합성곱과 계층적 다운샘플링으로 전역 정보를 “깊이”에서 얻게 되는 반면, ViT는 자기어텐션으로 처음부터 전역 관계를 직접 학습하는 것이 특징이다. 충분한 데이터와 적절한 사전학습·증강이 있을 때 스케일이 클수록 성능이 잘 향상되지만, 어텐션의 계산량이 토큰 수 제곱(O(N²))에 비례해 해상도가 높거나 패치가 너무 작으면 비용이 크게 늘 수 있다.

![ViT](./images/ViT.png)

> NLP transformer에서 단어(Token)가 들어가던 자리에 ViT에서는 이미지 patch token이 들어간다고 생각하면 된다.

1. NLP Transformer
- 문장을 Token으로 분리
- Token ID > Embedding
- [CLS] Token
- Position Embedding
- Multi-Head Self-Attention
- FFN/MLP
- 문장 분류

2. Vision Transformer
- 이미지를 Patch로 분리
- Patch > Linear Projection
- [CLS] Token
- Position Embedding
- Multi-Head Self-Attention
- FFN/MLP
- 이미지 분류

이미지가 (3, 224, 224)이고 patch 크기가 16 * 16이면 한 변에 224 / 16 = 14개의 patch가 생긴다. Patch 수는 14 * 14 = 196개이고 [CLS] token까지 추가하면 transformer가 받는 sequence 길이는 197이다. NLP의 Self-Attention과 계산 원리는 같다(각 patch가 Query, Key, Value를 만들고 다른 모든 patch와의 관계를 계산한다).
> NLP와 ViT의 차이는 토큰의 출처가 글자인가, 이미지 조각인가의 차이이다.

````python
from pathlib import Path
import json
import random
import math
import time
import os

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

SEED = 2026

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

if torch.cuda.is_available():
    DEVICE = torch.device('cuda')
elif torch.backends.mps.is_available():
    DEVICE = torch.device('mps')
elif torch.xpu.is_available():
    DEVICE = torch.device('xpu')
else:
    DEVICE = torch.device('cpu')

print(DEVICE)
````

````python
rng = np.random.default_rng(SEED)
image = rng.random((8, 8, 3), dtype=np.float32)

patch_size = 4
h, w, c = image.shape

# (H, W, C)
# (패치 세로 개수, patch_h, 패치 가로 개수, patch_w, C)
patches = image.reshape(h // patch_size, patch_size, w // patch_size, patch_size, c)

# patch의 위치가 앞쪽 차원에 오도록 변경
patches = patches.transpose(0, 2, 1, 3, 4)

# (총 patch 수, patch_h, patch_w, C)
patches = patches.reshape(-1, patch_size, patch_size, c)

print("원본 이미지:", image.shape)
print("patches:", patches.shape)
````

````python
fig, axes = plt.subplots(1, len(patches), figsize=(10, 3))

for i, ax in enumerate(axes):
    ax.imshow(patches[i])
    ax.set_title(f"Patch {i}")
    ax.axis("off")

plt.tight_layout()
plt.show()
````

### Conv2d가 Patch Embedding이 되는 이유
kernel_size=16, stride=16인 convolution은 서로 겹치지 않는 16 * 16 영역을 한 번씩 확인하는데, 출력 channel을 embed_dim으로 설정하면 각 patch를 embed_dim차원의 vector로 바꾸는 효과가 나게된다.
> Patch 자르기 > Flatten > Linear == Conv2d(kernel=patch_size, stride=patch_size)

````python
class PatchEmbedding(nn.Module):
    def __init__(self, patch_size=4, in_channels=3, enbed_dim=32):
        super().__init__()
        self.patch_size = patch_size
        self.proj = nn.Conv2d(in_channels, enbed_dim, kernel_size=patch_size, stride=patch_size)

    def forward(self, x):
        # x: (B, C, H, W)
        x = self.proj(x)
        # x: (B, D, H/P, W/P)
        x = x.flatten(2)
        # x: (B, D, N);N = H/P * W/P
        x = x.transpose(1, 2)
        # x: (B, N, D)
        return x
````

````python
sample = torch.from_numpy(image).permute(2, 0, 1).unsqueeze(0)# (1, 3, 8, 8)
patch_embed = PatchEmbedding(patch_size=4, in_channels=3, enbed_dim=32)
tokens = patch_embed(sample)
print("입력:", tokens)
print("토큰:", tokens.shape)
````

### [CLS] Token과 Position Embedding
원조 ViT 구조
- [CLS] token은 NLP의 BERT에서 사용하던 방식과 매우 유사하다. 모든 patch와 attention을 주고 받으면서 이미지 전체를 대표하는 표현으로 학습되고, 마지막에 [CLS] token을 classification head에 넣어준다.
> 최근 vision backbone은 [CLS] 하나만 사용하는 방식 외에도 global average pooling, register token, 상대/회전 위치 표현 등 여러 변형을 사용한다.

````python
# B: 이미지 개수
# N: 이미지 한 장을 4개의 patch로 나눔
# D: 각 patch를 32개의 숫자로 표현
B, N, D = tokens.shape
cls_token = nn.Parameter(torch.zeros(1, 1, D)) # nn.Parameter(): 학습되는 vector로 만듦
pos_embed = nn.Parameter(torch.zeros(1, N+1, D))
nn.init.trunc_normal_(cls_token, std=0.02) # trunc_normal_(): 절단 정규분포;아주 작은 값으로 초기화
nn.init.trunc_normal_(pos_embed, std=0.02)
# CLS shape: (1, 1, 32)
# 이미지가 2장인 CLS shape: (2, 1, 32)
x = torch.cat([cls_token.expand(B, -1, -1), tokens], dim=1)
x = x + pos_embed
print("Patch token 수:", N)
print("CLS 추가 후:", x.shape[1])
````

### Transformer Encoder에 넣기

````python
'''
입력 > LayerNorm > Multi-Head Self-Attention > FFN > 출력

D: 차원
nhead: Multi-Head Attention의 Head를 4개 사용 > 서로 다른 관계를 학습할 기회
dim_feedforward: 각 Token의 표현을 더 풍부하게 변환하는 역활
batch_first: (B, N, D) 형태 그대로
norm_first: Layer Norm을 Attention과 FFN보다 먼저 할 것인지를 설정
'''
encoder_layer = nn.TransformerEncoderLayer(
    d_model=D, nhead=4, dim_feedforward=D*4, dropout=0.1, activation="gelu",
    batch_first=True, norm_first=True
)

# norm=nn.LayerNorm(D): 최종 출력에 LayerNorm을 한 번 더 적용
encoder = nn.TransformerEncoder(encoder_layer, num_layers=2, norm=nn.LayerNorm(D))

z = encoder(x)
classifier = nn.Linear(D, 10)
logits = classifier(z[:, 0])

print("Encoder 출력: ", z.shape)
print("분류 logits: ", logits.shape)
print("예측 class: ", logits.argmax(dim=1).item())
````

하나의 이미지에서 4개의 숫자를 각각 예측하는 multi-head classification 문제
- 예시) 정답이 "1234", head1: 1, head2: 2, head3: 3, head4: 4

Transformer backbone은 하나를 공유하고, 마지막 classification head만 4개로 나눈다. Transformer 관점에서는 공통 encoder 표현을 여러 task head가 공유하는 multi-task learning과 비슷하다.

````python
from collections import Counter

position_counts = [Counter() for _ in range(4)]

for item in annotations:
    for pos, label in enumerate(item["labels"]):
        position_counts[pos][int(label)] += 1

for pos, counts in enumerate(position_counts, start=1):
    print(f"\n{pos}번째 자리 분포")
    print(dict(sorted(counts.items())))
````

````python
indices = np.arange(len(annotations))
rng = np.random.default_rng(SEED)
rng.shuffle(indices)

split = int(len(indices) * 0.9)
train_idx = indices[:split]
val_idx = indices[split:]

train_annot = [annotations[i] for i in train_idx]
val_annot = [annotations[i] for i in val_idx]

print("Train:", len(train_annot))
print("Val  :", len(val_annot))
````

````python
# 이미지 전처리와 데이터 증강 라이브러리
# ToTensorV2: 파이썬 Tensor 변환, OpenCV이미지 Shape: (H, W, C), 파이토치 모델이 원하는 Shape: (C, H, W)
import albumentations as A
from albumentations.pytorch import ToTensorV2

IMAGE_SIZE = 224

train_transform = A.Compose([
    A.Affine(
        scale=(0.92, 1.08),
        translate_percent=(-0.04, 0.04),
        rotate=(-8, 8),  # -8~8도 사이로 무작위 회전
        shear=(-3, 3),  # -3~3도 사이로 기울이기
        border_mode=0, # 빈공간을 검정색으로 채우기
        p=0.6,  # 60% 확률로만 이 변형을 적용
    ),
    A.RandomBrightnessContrast(
        brightness_limit=0.15,  # 밝기를 최대 ±15%까지 무작위로 변화
        contrast_limit=0.15,  # 대비를 최대 ±15%까지 무작위로 변화
        p=0.3,  # 30% 확률로만 적용
    ),
    A.LongestMaxSize(max_size=IMAGE_SIZE),  # 긴 변을 224로 맞춰서 비율 유지한 채 축소
    A.PadIfNeeded(
        min_height=IMAGE_SIZE,
        min_width=IMAGE_SIZE,
        border_mode=0,
    ),
    A.Normalize(
        mean=(0.485, 0.456, 0.406),  # ImageNet 사전학습 통계값으로 정규화
        std=(0.229, 0.224, 0.225),
    ),
    ToTensorV2(), # numpy 배열을 (C, H, W) 파이토치 텐서로 변환
])

val_transform = A.Compose([
    A.LongestMaxSize(max_size=IMAGE_SIZE),
    A.PadIfNeeded(
        min_height=IMAGE_SIZE,
        min_width=IMAGE_SIZE,
        border_mode=0,
    ),
    A.Normalize(
        mean=(0.485, 0.456, 0.406),
        std=(0.229, 0.224, 0.225),
    ),
    ToTensorV2(),
])
````

````python
class JsonDataset(Dataset):
    def __init__(self, image_dir, annotations, transform=None):
        self.image_dir = Path(image_dir)
        self.annotations = annotations
        self.transform = transform
        self.num_classes = 10

    def __len__(self):
        return len(self.annotations)

    def __getitem__(self, idx):
        item = self.annotations[idx]
        image_path = self.image_dir / item["filename"]
        image = np.array(Image.open(image_path).convert("RGB"))
        target = torch.tensor(
            [int(x) for x in item["labels"]],
            dtype=torch.long
        )

        if self.transform is not None:
            image = self.transform(image=image)["image"]

        return image, target
````

````python
train_dataset = JsonDataset(IMAGE_DIR, train_annot, train_transform)
val_dataset = JsonDataset(IMAGE_DIR, val_annot, val_transform)
````

````python
NUM_WORKERS = 0 if os.name == "nt" else min(4, os.cpu_count() or 1)
BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
    persistent_workers=(NUM_WORKERS > 0),
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
    persistent_workers=(NUM_WORKERS > 0),
)

images, targets = next(iter(train_loader))
print("images:", images.shape)
print("targets:", targets.shape)
print("예시 target:", targets[0])
````

```
- ViT-Base보다 가볍고 실습하기 좋다.
- 입력 데이터: 224 * 224, patch: 16 * 16
- Pretrained를 사용하므로 적은 데이터에서도 일반 학습보다 유리하다.
```

````python
import timm
````

````python
MODEL_NAME = "vit_small_patch16_224"
````

````python
class ViTMultiHeadClassifier(nn.Module):
    def __init__(self, num_classes=10, num_heads_out=4, model_name=MODEL_NAME, pretrained=True, head_dropout=0.2):
        super().__init__()

        # num_classes=0:
        # timm 모델의 원래 ImageNet classifier를 제거하고 feature만 받습니다.
        # timm 모델은 기존의 모델을 트렌드에 맞게 수정시켜서 올려 놓는 경우가 많다.
        self.backbone = timm.create_model(
            model_name,
            pretrained=pretrained, # 실제 학습된 가중치 사용
            num_classes=0,
        )

        feature_dim = self.backbone.num_features

        # 네 자리 숫자를 각각 예측하는 독립적인 classification head
        self.heads = nn.ModuleList([
            nn.Sequential(
                nn.LayerNorm(feature_dim),
                nn.Dropout(head_dropout),
                nn.Linear(feature_dim, num_classes),
            )
            for _ in range(num_heads_out)
        ])

    def forward(self, x):
        features = self.backbone(x)  # (B, feature_dim)
        return [head(features) for head in self.heads]


model = ViTMultiHeadClassifier().to(DEVICE)
print(model)
````

````python
sample_input = torch.randn(2, 3, IMAGE_SIZE, IMAGE_SIZE, device=DEVICE)
with torch.no_grad():
    sample_outputs = model(sample_input)

print("Head 개수:", len(sample_outputs))
for i, out in enumerate(sample_outputs, start=1):
    print(f"Head {i}:", out.shape)
````

````python
EPOCHS = 10
LR = 3e-5 # 0.00003
WEIGHT_DECAY = 0.05 # 과적합 방지
LABEL_SMOOTHING = 0.1 # 정답 class의 확률에 1.0을 몰아주는 대신 부드러운 target을 사용하여 과도한 확신을 줄이고 일반화를 돕게 함

criterion = nn.CrossEntropyLoss(label_smoothing=LABEL_SMOOTHING)
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

# CosineAnnealingLR(): cos 형태로 부드러운 곡선 형태로 감소하게 해줌
# T_max: 해당 숫자만큼 호출되도록 해줌
# eta_min: LR을 어디까지 낮출 것인지 설정
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=LR * 0.05)

print(optimizer)
````

### AMP(Automatic Mixed Precision)
GPU에서는 mixed precision을 사용하며, 연산 속도를 높이고 GPU 메모리 사용량을 줄일 수 있다.
- 자동 혼합 정밀도(AMP): 부동소수점 지원을 상황에 맞게 섞어 사용할 수 있게 함
    - FP32(32bit 부동소수점): 높은 정밀도, 더 많은 메모리, 계산량 많음
    - FP16(16bit 부동소수점): 낮은 정밀도, 메모리 절약, 계산량 적음
- torch.amp.autocast, torch.amp.GradScaler

````python
use_amp = (DEVICE.type == "cuda")
scaler = torch.amp.GradScaler("cuda", enabled=use_amp)

print("AMP enabled:", use_amp)
````

````python
@torch.no_grad()
def evaluate(model, loader, criterion, device):
    model.eval()

    total_loss = 0.0
    total_samples = 0
    correct_per_head = torch.zeros(4, dtype=torch.long)
    exact_correct = 0

    for images, targets in loader:
        images = images.to(device, non_blocking=True)
        targets = targets.to(device, non_blocking=True)

        with torch.amp.autocast(
            device_type=device.type,
            dtype=torch.float16,
            enabled=(device.type == "cuda"),
        ):
            outputs = model(images)
            losses = [
                criterion(outputs[i], targets[:, i])
                for i in range(4)
            ]
            loss = torch.stack(losses).mean()

        batch_size = images.size(0)
        total_loss += loss.item() * batch_size
        total_samples += batch_size

        preds = torch.stack(
            [out.argmax(dim=1) for out in outputs],
            dim=1
        )  # (B, 4)

        correct_matrix = preds.eq(targets)

        correct_per_head += correct_matrix.sum(dim=0).cpu()
        exact_correct += correct_matrix.all(dim=1).sum().item()

    digit_acc = (correct_per_head.float() / total_samples).tolist()
    exact_acc = exact_correct / total_samples

    return {
        "loss": total_loss / total_samples,
        "digit_acc": digit_acc,
        "exact_acc": exact_acc,
    }
````

````python
def train_one_epoch(model, loader, optimizer, criterion, scaler, device):
    model.train()

    total_loss = 0.0
    total_samples = 0

    for images, targets in loader:
        images = images.to(device, non_blocking=True)
        targets = targets.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast(
            device_type=device.type,
            dtype=torch.float16,
            enabled=(device.type == "cuda"),
        ):
            outputs = model(images)

            losses = [
                criterion(outputs[i], targets[:, i])
                for i in range(4)
            ]

            loss = torch.stack(losses).mean()

        scaler.scale(loss).backward()

        # clipping 전에 unscale
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

        scaler.step(optimizer)
        scaler.update()

        batch_size = images.size(0)
        total_loss += loss.item() * batch_size
        total_samples += batch_size

    return total_loss / total_samples
````

````python
history = {
    "train_loss": [],
    "val_loss": [],
    "exact_acc": [],
    "digit_acc": [],
}

best_exact_acc = -1.0
best_path = MODEL_DIR / "best_vit_multihead.pth"

for epoch in range(1, EPOCHS + 1):
    start = time.time()

    train_loss = train_one_epoch(
        model,
        train_loader,
        optimizer,
        criterion,
        scaler,
        DEVICE,
    )

    metrics = evaluate(
        model,
        val_loader,
        criterion,
        DEVICE,
    )

    scheduler.step()

    history["train_loss"].append(train_loss)
    history["val_loss"].append(metrics["loss"])
    history["exact_acc"].append(metrics["exact_acc"])
    history["digit_acc"].append(metrics["digit_acc"])

    digit_text = " | ".join(
        f"D{i+1}: {acc*100:6.2f}%"
        for i, acc in enumerate(metrics["digit_acc"])
    )

    print(
        f"[{epoch:02d}/{EPOCHS}] "
        f"train_loss={train_loss:.4f} | "
        f"val_loss={metrics['loss']:.4f} | "
        f"{digit_text} | "
        f"Exact={metrics['exact_acc']*100:6.2f}% | "
        f"lr={optimizer.param_groups[0]['lr']:.2e} | "
        f"{time.time()-start:.1f}s"
    )

    if metrics["exact_acc"] > best_exact_acc:
        best_exact_acc = metrics["exact_acc"]

        torch.save({
            "epoch": epoch,
            "model_name": MODEL_NAME,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "best_exact_acc": best_exact_acc,
            "image_size": IMAGE_SIZE,
            "num_classes": 10,
            "num_heads_out": 4,
        }, best_path)

        print(f"  -> Best model 저장: {best_path}")

print(f"\nBest Exact Match Accuracy: {best_exact_acc*100:.2f}%")
````

````python
from torchvision.transforms.functional import to_pil_image

IMAGENET_MEAN = torch.tensor(
    [0.485, 0.456, 0.406]
).view(3, 1, 1)

IMAGENET_STD = torch.tensor(
    [0.229, 0.224, 0.225]
).view(3, 1, 1)

def tensor_to_pil(tensor):
    tensor = tensor.detach().cpu()
    tensor = tensor * IMAGENET_STD + IMAGENET_MEAN
    tensor = tensor.clamp(0, 1)
    return to_pil_image(tensor)

def digits_to_text(x):
    return "".join(str(int(v)) for v in x)
````

````python
@torch.no_grad()
def show_predictions(model, loader, device, max_images=8):
    model.eval()

    shown = 0
    rows = math.ceil(max_images / 4)
    fig, axes = plt.subplots(rows, 4, figsize=(12, 3 * rows))
    axes = np.array(axes).reshape(-1)

    for images, targets in loader:
        images_device = images.to(device)

        outputs = model(images_device)
        preds = torch.stack(
            [out.argmax(dim=1) for out in outputs],
            dim=1
        ).cpu()

        for i in range(images.size(0)):
            if shown >= max_images:
                break

            true_text = digits_to_text(targets[i])
            pred_text = digits_to_text(preds[i])

            axes[shown].imshow(tensor_to_pil(images[i]))
            axes[shown].set_title(
                f"GT: {true_text} / Pred: {pred_text}"
            )
            axes[shown].axis("off")
            shown += 1

        if shown >= max_images:
            break

    for ax in axes[shown:]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()

show_predictions(best_model, val_loader, DEVICE)
````
